## Filtering with conditions

```python
import pandas as pd

txns = pd.DataFrame({
    "txn_id": [1, 2, 3, 4, 5],
    "customer": ["Priya", "Rahul", "Anita", "Priya", "Rahul"],
    "type": ["debit", "credit", "debit", "debit", "credit"],
    "amount": [1200.0, 50000.0, 349.0, 15000.0, 2000.0]
})
big = txns[txns["amount"] > 10000]
print(big)
big_debits = txns[(txns["amount"] > 1000) & (txns["type"] == "debit")]
print(big_debits)
```

apply, map and groupby

```python
import pandas as pd

emp = pd.DataFrame({
    "name": ["Priya Sharma", "Rahul Verma", "Anita Desai"],
    "dept_code": ["DS", "FIN", "HR"],
    "salary": [88000, 65000, 61000]
})
# map: replace codes using a dictionary (a lookup table)
dept_names = {"DS": "Data Science", "FIN": "Finance", "HR": "Human Resources"}
emp["department"] = emp["dept_code"].map(dept_names)
# apply: run a custom function on every value of a column
def tax_slab(salary):
    return "high" if salary > 70000 else "standard"

emp["tax_slab"] = emp["salary"].apply(tax_slab)
print(emp[["name", "department", "tax_slab"]])
```

2.groupby: split, apply, combine

```python
import pandas as pd

orders = pd.DataFrame({
    "city": ["Mumbai", "Delhi", "Mumbai", "Pune", "Delhi", "Mumbai"],
    "amount": [2499.0, 1250.5, 799.0, 3999.0, 5400.0, 1500.0]
})
print(orders.groupby('city')['amount'].sum())
```

3. several aggregations at once

```python
import pandas as pd

orders = pd.DataFrame({
    "city": ["Mumbai", "Delhi", "Mumbai", "Pune", "Delhi", "Mumbai"],
    "amount": [2499.0, 1250.5, 799.0, 3999.0, 5400.0, 1500.0]
})
reports = orders.groupby('city')['amount'].agg(['count', 'sum', 'mean'])
print(reports)

report2 = orders.groupby("city").agg(
    total_sales=("amount", "sum"),
    biggest_order=("amount", "max")
)
print(report2)
```

SafeBank: branches ["Andheri", "Karol Bagh", "Andheri", "Karol Bagh"], types ["debit", "credit", "debit", "debit"], amounts [1200.0, 50000.0, 349.0, 15000.0]. Add a fee column using apply with a lambda (Day 15): 10.0 if amount > 10000 else 0.0. Then print count and sum of amount per branch.

```python
import pandas as pd

txns = pd.DataFrame({
    "branch": ["Andheri", "Karol Bagh", "Andheri", "Karol Bagh"],
    "type": ["debit", "credit", "debit", "debit"],
    "amount": [1200.0, 50000.0, 349.0, 15000.0]
})
txns['fee'] = txns['amount'].apply(lambda a: 10.0 if a > 1000 else 0.0)
print(txns.groupby("branch")["amount"].agg(["count", "sum"])) 
```

```python
import pandas as pd

customers = pd.DataFrame({
    "cust_id": [1, 2, 3],
    "name": ["Priya Sharma", "Rahul Verma", "Anita Desai"],
    "city": ["Mumbai", "Delhi", "Pune"]
})
orders = pd.DataFrame({
    "order_id": [101, 102, 103, 104],
    "cust_id": [1, 2, 1, 4],
    "amount": [2499.0, 1250.5, 799.0, 999.0]
})

inner = pd.merge(orders, customers, on='cust_id')
print(inner)
```

Left merge: keep every order

```python
import pandas as pd

customers = pd.DataFrame({
    "cust_id": [1, 2, 3],
    "name": ["Priya Sharma", "Rahul Verma", "Anita Desai"]
})
orders = pd.DataFrame({
    "order_id": [101, 102, 104],
    "cust_id": [1, 2, 4],
    "amount": [2499.0, 1250.5, 999.0]
})

left = pd.merge(orders, customers, on='cust_id', how='left')
print(left)
```
